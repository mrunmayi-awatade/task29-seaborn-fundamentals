```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    "Department": ["IT", "HR", "IT", "Sales", "HR", "Sales", "IT", "HR"],
    "Salary": [45000, 35000, 50000, 40000, 38000, 42000, 55000, 36000],
    "Experience": [2, 3, 5, 4, 2, 3, 6, 4],
    "Gender": ["Male", "Female", "Male", "Female", "Female", "Male", "Female", "Male"]
}
df = pd.DataFrame(data)
print(df)
```

```python
sns.countplot(data=df, x="Department")
plt.title("Employee Count by Department")
plt.xlabel("Department")
plt.ylabel("Count")
plt.show()
```

```python
sns.boxplot(data=df, x="Department", y="Salary")
plt.title("Salary Distribution by Department")
plt.xlabel("Department")
plt.ylabel("Salary")
plt.show()
```

```python
sns.scatterplot(data=df, x="Experience", y="Salary", hue="Department")
plt.title("Experience vs Salary")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary")
plt.show()
```

```python
sns.histplot(data=df, x="Salary", bins=5, kde=True)
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.violinplot(data=df, x="Department", y="Salary")
plt.title("Salary Distribution by Department")
plt.xlabel("Department")
plt.ylabel("Salary")
plt.show()
```

```python
numeric_df = df[["Salary", "Experience"]]
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm")
plt.title("Salary and Experience Correlation")
plt.show()
```

```python
plt.figure(figsize=(8, 4))
plt.subplot(1, 2, 1)
sns.barplot(data=df, x="Department", y="Salary", errorbar=None)
plt.title("Seaborn Bar Plot")
plt.subplot(1, 2, 2)
plt.bar(df["Department"], df["Salary"])
plt.title("Matplotlib Bar Plot")
plt.tight_layout()
plt.show()
```

```python
sns.set_style("whitegrid")
sns.barplot(data=df, x="Department", y="Salary", errorbar=None)
plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.show()
```

```python
sns.pairplot(df, vars=["Salary", "Experience"], hue="Department")
plt.show()
```

```python
sns.barplot(data=df, x="Department", y="Salary", errorbar=None)
plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.savefig("seaborn_barplot.png")
plt.show()
```
